```python
#Variance Inflation Factor
X = telecom.copy().drop(
    columns=["churn", "Customer_ID"],
    errors="ignore"
)

# 2. Select numerical features
X = X.select_dtypes(include=np.number)

# 3. Handle infinite and missing values
X = X.replace([np.inf, -np.inf], np.nan)
X = X.fillna(X.median())

# 4. Remove constant features
X = X.loc[:, X.nunique() > 1]

# 5. Add intercept
X_vif = add_constant(X, has_constant="add")

# 6. Calculate VIF
vif_results = pd.DataFrame({
    "Feature": X_vif.columns,
    "VIF": [
        variance_inflation_factor(X_vif.values, i)
        for i in range(X_vif.shape[1])
    ]
})

# 7. Remove intercept and sort
vif_results = (
    vif_results[vif_results["Feature"] != "const"]
    .sort_values(by="VIF", ascending=False)
    .reset_index(drop=True)
)


plt.figure(figsize=(12, 16))

sns.barplot(
    data=vif_results.sort_values("VIF", ascending=True),
    x="VIF",
    y="Feature",
    color="steelblue"
)

plt.axvline(5, color="orange", linestyle="--", label="VIF = 5")
plt.axvline(10, color="red", linestyle="--", label="VIF = 10")

plt.title("Variance Inflation Factor — Telecom Dataset")
plt.xlabel("VIF")
plt.ylabel("Features")
plt.legend()
plt.tight_layout()
plt.savefig("Visualizations/VIF.png")
plt.show()
```

```python
voice_features = [
    "plcd_vce_Mean",
    "attempt_Mean",
    "comp_vce_Mean",
    "complete_Mean"
]

X_features = telecom.drop(
    columns=["churn", "Customer_ID"],
    errors="ignore"
).select_dtypes(include=np.number)

X_features = (
    X_features
    .replace([np.inf, -np.inf], np.nan)
    .fillna(X_features.median())
)

X_features = X_features.loc[:, X_features.nunique() > 1]

def matrix_diagnostics(data):
    scaled = StandardScaler().fit_transform(data)

    singular_values = np.linalg.svd(
        scaled,
        compute_uv=False
    )

    rank = np.linalg.matrix_rank(scaled)
    condition_number = (
        singular_values[0] / singular_values[-1]
    )

    return {
        "Features": data.shape[1],
        "Rank": rank,
        "Smallest_Singular_Value": singular_values[-1],
        "Condition_Number": condition_number
    }

baseline = matrix_diagnostics(X_features)

results = []

for feature in voice_features:
    reduced = X_features.drop(columns=feature)

    diagnostics = matrix_diagnostics(reduced)
    diagnostics["Removed_Feature"] = feature

    results.append(diagnostics)

ablation_results = pd.DataFrame(results)

print("Baseline:")
print(baseline)

print("\nSingle-feature removal:")
print(
    ablation_results.sort_values(
        "Condition_Number"
    ).to_string(index=False)
) 
```

```python
X_features = (
    telecom
    .drop(columns=["churn", "Customer_ID"], errors="ignore")
    .select_dtypes(include=np.number)
    .replace([np.inf, -np.inf], np.nan)
)

# Median imputation for EDA diagnostics only
X_features = X_features.fillna(X_features.median())

# Remove constant features
X_features = X_features.loc[:, X_features.nunique() > 1]

# Standardize
scaler = StandardScaler()
X_std = scaler.fit_transform(X_features)

# ============================================================
# 2. SINGULAR VALUE DECOMPOSITION
# ============================================================

U, S, Vt = np.linalg.svd(
    X_std,
    full_matrices=False
)

print("=" * 70)
print("SINGULAR VALUE SPECTRUM")
print("=" * 70)

print("\n10 Smallest Singular Values:")
print(S[-10:])

print("\n10 Smallest Relative Singular Values:")
print(S[-10:] / S[0])

print("\nCondition Number:")
print(S[0] / S[-1])

# ============================================================
# 3. NUMERICAL RANK SENSITIVITY
# ============================================================

print("\n" + "=" * 70)
print("NUMERICAL RANK SENSITIVITY")
print("=" * 70)

for rtol in [1e-8, 1e-9, 1e-10, 1e-11, 1e-12]:
    rank = np.sum(S > S[0] * rtol)

    print(
        f"Relative tolerance: {rtol:.0e} | "
        f"Rank: {rank}/{len(S)} | "
        f"Deficiency: {len(S) - rank}"
    )

# ============================================================
# 4. INSPECT FOUR SMALLEST SINGULAR DIRECTIONS
# ============================================================

print("\n" + "=" * 70)
print("NEAR-NULL SINGULAR VECTORS")
print("=" * 70)

near_null_results = []

for i in range(1, 5):

    vector = pd.Series(
        Vt[-i],
        index=X_features.columns,
        name=f"Direction_{i}"
    )

    top_features = vector.abs().nlargest(15).index

    result = pd.DataFrame({
        "Feature": top_features,
        "Coefficient": vector.loc[top_features].values,
        "Absolute_Coefficient": vector.loc[top_features].abs().values
    })

    near_null_results.append(result)

    print(f"\nDirection {i}")
    print(f"Singular value: {S[-i]:.12e}")
    print(result.to_string(index=False))

# ============================================================
# 5. FEATURE VARIANCE DIAGNOSTICS
# ============================================================

print("\n" + "=" * 70)
print("LOW-VARIANCE FEATURE CHECK")
print("=" * 70)

feature_std = X_features.std()

print("\nZero-variance features:")
print(feature_std[feature_std == 0])

print("\n10 Lowest Standard Deviations:")
print(feature_std.nsmallest(10))

# ============================================================
# 6. CORRELATION STRUCTURE
# ============================================================

print("\n" + "=" * 70)
print("HIGH-CORRELATION FEATURE PAIRS")
print("=" * 70)

corr_matrix = X_features.corr().abs()

upper_triangle = corr_matrix.where(
    np.triu(
        np.ones(corr_matrix.shape),
        k=1
    ).astype(bool)
)

correlation_pairs = (
    upper_triangle
    .stack()
    .reset_index()
)

correlation_pairs.columns = [
    "Feature_1",
    "Feature_2",
    "Absolute_Correlation"
]

correlation_pairs = correlation_pairs.sort_values(
    "Absolute_Correlation",
    ascending=False
)

print(
    correlation_pairs.head(30).to_string(index=False)
)

# ============================================================
# 7. VOICE-CALL FEATURE GROUP ABLATION
# ============================================================

print("\n" + "=" * 70)
print("VOICE-CALL GROUP ABLATION")
print("=" * 70)

voice_features = [
    "plcd_vce_Mean",
    "attempt_Mean",
    "comp_vce_Mean",
    "complete_Mean"
]

voice_features = [
    f for f in voice_features
    if f in X_features.columns
]

ablation_results = []

for k in range(1, len(voice_features) + 1):

    for subset in combinations(voice_features, k):

        reduced = X_features.drop(
            columns=list(subset)
        )

        reduced_std = StandardScaler().fit_transform(reduced)

        singular_values = np.linalg.svd(
            reduced_std,
            compute_uv=False
        )

        ablation_results.append({
            "Removed_Features": ", ".join(subset),
            "N_Removed": k,
            "Remaining_Features": reduced.shape[1],
            "Rank": np.linalg.matrix_rank(reduced_std),
            "Smallest_Singular_Value": singular_values[-1],
            "Condition_Number": (
                singular_values[0] / singular_values[-1]
            )
        })

ablation_results = pd.DataFrame(ablation_results)

print(
    ablation_results
    .sort_values("Condition_Number")
    .to_string(index=False)
)

# ============================================================
# 8. FINAL DIAGNOSTIC SUMMARY
# ============================================================

print("\n" + "=" * 70)
print("FINAL MULTICOLLINEARITY SUMMARY")
print("=" * 70)

print(f"Observations: {X_std.shape[0]:,}")
print(f"Features: {X_std.shape[1]}")
print(f"Default numerical rank: {np.linalg.matrix_rank(X_std)}")
print(f"Smallest singular value: {S[-1]:.12e}")
print(f"Largest singular value: {S[0]:.12e}")
print(f"Condition number: {S[0] / S[-1]:.12e}")

print("\nNear-null directions identified: 4")
print("Feature removal: NOT performed")
print("Original telecom dataframe: NOT modified")
```

```python

feature_groups = {
    "Voice_Call": [
        "plcd_vce_Mean",
        "attempt_Mean",
        "comp_vce_Mean",
        "complete_Mean",
        "peak_vce_Mean",
        "opk_vce_Mean"
    ],

    "Data_Call": [
        "plcd_dat_Mean",
        "comp_dat_Mean",
        "blck_dat_Mean",
        "drop_dat_Mean",
        "drop_blk_Mean",
        "unan_dat_Mean",
        "opk_dat_Mean",
        "peak_dat_Mean"
    ],

    "Dropped_Blocked_Calls": [
        "drop_blk_Mean",
        "blck_vce_Mean",
        "drop_vce_Mean",
        "blck_dat_Mean",
        "drop_dat_Mean"
    ]
}

results = []

for group_name, features in feature_groups.items():

    features = [
        f for f in features
        if f in X_features.columns
    ]

    group_data = X_features[features]

    for target in features:

        predictors = [
            f for f in features
            if f != target
        ]

        if not predictors:
            continue

        X_group = group_data[predictors]
        y_group = group_data[target]

        model = LinearRegression()
        model.fit(X_group, y_group)

        predictions = model.predict(X_group)
        residuals = y_group.to_numpy() - predictions

        results.append({
            "Group": group_name,
            "Target": target,
            "Predictors": ", ".join(predictors),
            "R2": model.score(X_group, y_group),
            "Residual_STD": np.std(residuals),
            "Residual_Median_Abs": np.median(np.abs(residuals)),
            "Residual_P99_Abs": np.quantile(
                np.abs(residuals), 0.99
            ),
            "Residual_Max_Abs": np.max(np.abs(residuals))
        })

reconstruction_results = pd.DataFrame(results)

print("=" * 90)
print("FEATURE RECONSTRUCTION RESULTS")
print("=" * 90)

for group_name in feature_groups:

    print(f"\n{group_name}")
    print("-" * 90)

    group_results = (
        reconstruction_results[
            reconstruction_results["Group"] == group_name
        ]
        .sort_values("R2", ascending=False)
    )

    print(
        group_results[
            [
                "Target",
                "R2",
                "Residual_STD",
                "Residual_Median_Abs",
                "Residual_P99_Abs",
                "Residual_Max_Abs"
            ]
        ].to_string(index=False)
    )

print("\n" + "=" * 90)
print("MOST RECONSTRUCTIBLE FEATURES")
print("=" * 90)

print(
    reconstruction_results
    .sort_values("R2", ascending=False)
    .head(15)
    .to_string(index=False)
)
```

```python
feature_groups = {
    "Voice_Call": [
        "plcd_vce_Mean",
        "attempt_Mean",
        "comp_vce_Mean",
        "complete_Mean",
        "peak_vce_Mean",
        "opk_vce_Mean"
    ],

    "Data_Call": [
        "plcd_dat_Mean",
        "comp_dat_Mean",
        "blck_dat_Mean",
        "drop_dat_Mean",
        "drop_blk_Mean",
        "unan_dat_Mean",
        "opk_dat_Mean",
        "peak_dat_Mean"
    ],

    "Dropped_Blocked_Calls": [
        "drop_blk_Mean",
        "blck_vce_Mean",
        "drop_vce_Mean",
        "blck_dat_Mean",
        "drop_dat_Mean"
    ]
}

# ============================================================
# 1. GROUP-SPECIFIC SVD AND NUMERICAL RANK
# ============================================================

print("=" * 90)
print("GROUP-SPECIFIC SINGULAR VALUE ANALYSIS")
print("=" * 90)

for group_name, features in feature_groups.items():

    features = [
        f for f in features
        if f in X_features.columns
    ]

    group_matrix = StandardScaler().fit_transform(
        X_features[features]
    )

    singular_values = np.linalg.svd(
        group_matrix,
        compute_uv=False
    )

    print(f"\n{group_name}")
    print("-" * 90)

    print(f"Features: {len(features)}")
    print(
        "Rank at relative tolerance 1e-10:",
        np.sum(singular_values > singular_values[0] * 1e-10)
    )

    print("Singular values:")
    print(singular_values)

    print("Relative singular values:")
    print(singular_values / singular_values[0])

# ============================================================
# 2. STANDARDIZED RECONSTRUCTION COEFFICIENTS
# ============================================================

print("\n" + "=" * 90)
print("STANDARDIZED RECONSTRUCTION COEFFICIENTS")
print("=" * 90)

for group_name, features in feature_groups.items():

    features = [
        f for f in features
        if f in X_features.columns
    ]

    group_data = X_features[features]

    print(f"\n{group_name}")
    print("-" * 90)

    for target in features:

        predictors = [
            f for f in features
            if f != target
        ]

        X_group = group_data[predictors]
        y_group = group_data[target]

        X_scaled = StandardScaler().fit_transform(X_group)
        y_scaled = StandardScaler().fit_transform(
            y_group.to_numpy().reshape(-1, 1)
        ).ravel()

        model = LinearRegression()
        model.fit(X_scaled, y_scaled)

        coefficients = pd.Series(
            model.coef_,
            index=predictors
        )

        print(f"\nTarget: {target}")
        print(f"R²: {model.score(X_scaled, y_scaled):.12f}")

        print(
            coefficients
            .sort_values(key=abs, ascending=False)
            .to_string()
        )

# ============================================================
# 3. PAIRWISE CORRELATION WITHIN EACH GROUP
# ============================================================

print("\n" + "=" * 90)
print("WITHIN-GROUP CORRELATION STRUCTURE")
print("=" * 90)

for group_name, features in feature_groups.items():

    features = [
        f for f in features
        if f in X_features.columns
    ]

    print(f"\n{group_name}")
    print("-" * 90)

    print(
        X_features[features]
        .corr()
        .round(5)
        .to_string()
    )
```

```python
target_col = "churn"
id_cols = ["Customer_ID"]

feature_groups = {
    "Voice_Call": [
        "plcd_vce_Mean",
        "attempt_Mean",
        "comp_vce_Mean",
        "complete_Mean",
        "peak_vce_Mean",
        "opk_vce_Mean"
    ],

    "Data_Call": [
        "plcd_dat_Mean",
        "comp_dat_Mean",
        "blck_dat_Mean",
        "drop_dat_Mean",
        "drop_blk_Mean",
        "unan_dat_Mean",
        "opk_dat_Mean",
        "peak_dat_Mean"
    ],

    "Dropped_Blocked_Calls": [
        "drop_blk_Mean",
        "blck_vce_Mean",
        "drop_vce_Mean",
        "blck_dat_Mean",
        "drop_dat_Mean"
    ],

    "Overage": [
        "ovrrev_Mean",
        "vceovr_Mean",
        "ovrmou_Mean"
    ],

    "Customer_Care": [
        "ccrndmou_Mean",
        "cc_mou_Mean"
    ]
}

# ============================================================
# 2. PREPARE NUMERICAL DATA
# ============================================================

numeric_cols = telecom.select_dtypes(
    include=[np.number]
).columns.tolist()

excluded = [target_col] + id_cols

feature_cols = [
    col for col in numeric_cols
    if col not in excluded
]

X = telecom[feature_cols].replace(
    [np.inf, -np.inf], np.nan
).copy()

# Median imputation for diagnostic purposes only
X = X.fillna(X.median())

# Remove constant or invalid columns
valid_cols = [
    col for col in X.columns
    if X[col].nunique(dropna=False) > 1
]

X = X[valid_cols]

# Standardize
means = X.mean()
stds = X.std(ddof=0)

stds = stds.replace(0, 1)

Xz = ((X - means) / stds).to_numpy(dtype=np.float64)

column_index = {
    col: i for i, col in enumerate(valid_cols)
}

# Keep only available features in each group
groups = {
    group: [col for col in cols if col in column_index]
    for group, cols in feature_groups.items()
}

groups = {
    group: cols for group, cols in groups.items()
    if len(cols) > 0
}

print("Rows:", Xz.shape[0])
print("Features:", Xz.shape[1])

# ============================================================
# 3. CHECK GROUP OVERLAP
# ============================================================

print("\n" + "=" * 90)
print("GROUP OVERLAP")
print("=" * 90)

for group_a, group_b in combinations(groups.keys(), 2):

    overlap = sorted(
        set(groups[group_a]) & set(groups[group_b])
    )

    if overlap:
        print(f"{group_a} <-> {group_b}: {overlap}")

# ============================================================
# 4. CROSS-GROUP CORRELATION ANALYSIS
# ============================================================

print("\n" + "=" * 90)
print("CROSS-GROUP CORRELATIONS")
print("=" * 90)

cross_correlations = []

for group_a, group_b in combinations(groups.keys(), 2):

    features_a = groups[group_a]
    features_b = groups[group_b]

    for feature_a in features_a:
        for feature_b in features_b:

            if feature_a == feature_b:
                continue

            corr = X[feature_a].corr(X[feature_b])

            cross_correlations.append({
                "Group_A": group_a,
                "Feature_A": feature_a,
                "Group_B": group_b,
                "Feature_B": feature_b,
                "Correlation": corr,
                "Absolute_Correlation": abs(corr)
            })

cross_corr_df = pd.DataFrame(cross_correlations)

if not cross_corr_df.empty:

    cross_corr_df = cross_corr_df.sort_values(
        "Absolute_Correlation",
        ascending=False
    )

    print(
        cross_corr_df.head(40).to_string(index=False)
    )

# ============================================================
# 5. CROSS-GROUP LINEAR RECONSTRUCTION
# ============================================================

print("\n" + "=" * 90)
print("CROSS-GROUP RECONSTRUCTION")
print("=" * 90)

reconstruction_results = []

for target_group, target_features in groups.items():

    # Pool features from all other groups
    other_features = []

    for group_name, group_features in groups.items():

        if group_name != target_group:
            other_features.extend(group_features)

    # Deduplicate and exclude the target group's own features
    other_features = list(dict.fromkeys(other_features))

    other_features = [
        col for col in other_features
        if col not in target_features
        and col in column_index
    ]

    if not other_features:
        continue

    predictor_idx = [
        column_index[col] for col in other_features
    ]

    A = Xz[:, predictor_idx]

    for target in target_features:

        y = Xz[:, column_index[target]]

        # Intercept included explicitly
        A_design = np.column_stack([
            np.ones(A.shape[0]),
            A
        ])

        coefficients, residual_sum, rank, singular_values = (
            np.linalg.lstsq(A_design, y, rcond=1e-10)
        )

        predictions = A_design @ coefficients
        residuals = y - predictions

        ss_res = np.sum(residuals ** 2)
        ss_total = np.sum((y - y.mean()) ** 2)

        r2 = 1 - ss_res / ss_total if ss_total > 0 else np.nan

        rmse = np.sqrt(np.mean(residuals ** 2))
        mae = np.mean(np.abs(residuals))

        normalized_rmse = rmse / (
            np.std(y) + 1e-12
        )

        result = {
            "Target_Group": target_group,
            "Target": target,
            "Predictor_Count": len(other_features),
            "Predictor_Rank": rank,
            "R2": r2,
            "RMSE": rmse,
            "MAE": mae,
            "Normalized_RMSE": normalized_rmse,
            "Max_Abs_Error": np.max(np.abs(residuals))
        }

        reconstruction_results.append(result)

reconstruction_df = pd.DataFrame(
    reconstruction_results
)

if not reconstruction_df.empty:

    reconstruction_df = reconstruction_df.sort_values(
        "Normalized_RMSE"
    )

    print(
        reconstruction_df.to_string(index=False)
    )

# ============================================================
# 6. GLOBAL SVD: FOUR SMALLEST SINGULAR DIRECTIONS
# ============================================================

print("\n" + "=" * 90)
print("GLOBAL SVD: SMALLEST SINGULAR DIRECTIONS")
print("=" * 90)

# full_matrices=False avoids creating an unnecessarily large U matrix
U, singular_values, Vt = np.linalg.svd(
    Xz,
    full_matrices=False
)

relative_singular_values = (
    singular_values / singular_values[0]
)

print("\nSmallest 10 singular values:")
print(singular_values[-10:])

print("\nSmallest 10 relative singular values:")
print(relative_singular_values[-10:])

# Inspect the four smallest right singular vectors
for direction in range(1, 5):

    vector = Vt[-direction]

    loadings = pd.DataFrame({
        "Feature": valid_cols,
        "Loading": vector,
        "Absolute_Loading": np.abs(vector)
    }).sort_values(
        "Absolute_Loading",
        ascending=False
    )

    print("\n" + "-" * 90)
    print(
        f"NEAR-NULL DIRECTION {direction} | "
        f"Singular value: {singular_values[-direction]:.12e}"
    )
    print("-" * 90)

    print(loadings.head(15).to_string(index=False))

    # Summarize absolute loading contribution by group
    group_contributions = {}

    for group_name, group_features in groups.items():

        indices = [
            column_index[col]
            for col in group_features
            if col in column_index
        ]

        contribution = np.sum(vector[indices] ** 2)

        group_contributions[group_name] = contribution

    contribution_df = pd.Series(
        group_contributions,
        name="Squared_Loading_Contribution"
    ).sort_values(ascending=False)

    print("\nGroup-level squared loading contributions:")
    print(contribution_df.to_string())

# ============================================================
# 7. FINAL SUMMARY
# ============================================================

print("\n" + "=" * 90)
print("CROSS-GROUP ANALYSIS SUMMARY")
print("=" * 90)

print("\nTop cross-group correlations:")
print(
    cross_corr_df.head(15).to_string(index=False)
    if not cross_corr_df.empty else "No correlations available."
)

print("\nMost reconstructible targets using other groups:")
print(
    reconstruction_df.head(15).to_string(index=False)
    if not reconstruction_df.empty else "No reconstruction results."
)
```

```python
feature_groups = {
    "Voice_Call": [
        "plcd_vce_Mean",
        "attempt_Mean",
        "comp_vce_Mean",
        "complete_Mean",
        "peak_vce_Mean",
        "opk_vce_Mean"
    ],

    "Data_Call": [
        "plcd_dat_Mean",
        "comp_dat_Mean",
        "blck_dat_Mean",
        "drop_dat_Mean",
        "drop_blk_Mean",
        "unan_dat_Mean",
        "opk_dat_Mean",
        "peak_dat_Mean"
    ],

    "Dropped_Blocked_Calls": [
        "drop_blk_Mean",
        "blck_vce_Mean",
        "drop_vce_Mean",
        "blck_dat_Mean",
        "drop_dat_Mean"
    ]
}

MAX_SUBSET_SIZE = 3
TOP_RESULTS_PER_TARGET = 3
RANK_TOLERANCE = 1e-10

# ============================================================
# 2. PREPARE DATA
# ============================================================

available_groups = {
    group: [
        col for col in features
        if col in telecom.columns
    ]
    for group, features in feature_groups.items()
}

available_groups = {
    group: features
    for group, features in available_groups.items()
    if len(features) >= 2
}

all_features = list(dict.fromkeys(
    col
    for features in available_groups.values()
    for col in features
))

X = telecom[all_features].replace(
    [np.inf, -np.inf], np.nan
).copy()

# Median imputation for diagnostic purposes
X = X.fillna(X.median())

# Standardize
means = X.mean()
stds = X.std(ddof=0).replace(0, 1)

Xz = ((X - means) / stds).astype(np.float64)

# ============================================================
# 3. LINEAR MODEL EVALUATION FUNCTION
# ============================================================

def evaluate_linear_identity(y, A, predictor_names):

    # Include intercept
    A_design = np.column_stack([
        np.ones(A.shape[0]),
        A
    ])

    coefficients, residual_sum, rank, singular_values = (
        np.linalg.lstsq(
            A_design,
            y,
            rcond=RANK_TOLERANCE
        )
    )

    predictions = A_design @ coefficients
    residuals = y - predictions

    ss_res = np.sum(residuals ** 2)
    ss_total = np.sum((y - y.mean()) ** 2)

    r2 = (
        1 - ss_res / ss_total
        if ss_total > 0 else np.nan
    )

    rmse = np.sqrt(np.mean(residuals ** 2))
    mae = np.mean(np.abs(residuals))

    normalized_rmse = rmse / (
        np.std(y) + 1e-12
    )

    tolerance = 1e-8 * max(
        np.max(np.abs(y)), 1.0
    )

    exact_fraction = np.mean(
        np.abs(residuals) <= tolerance
    )

    coefficient_map = {
        "Intercept": coefficients[0]
    }

    coefficient_map.update({
        name: coef
        for name, coef in zip(
            predictor_names,
            coefficients[1:]
        )
    })

    return {
        "R2": r2,
        "RMSE": rmse,
        "MAE": mae,
        "Normalized_RMSE": normalized_rmse,
        "Max_Abs_Error": np.max(np.abs(residuals)),
        "Exact_Row_Fraction": exact_fraction,
        "Rank": rank,
        "Coefficients": coefficient_map
    }

# ============================================================
# 4. EXHAUSTIVE SPARSE LINEAR IDENTITY SEARCH
# ============================================================

all_results = []
best_models = []

for group_name, features in available_groups.items():

    print("\n" + "=" * 100)
    print("GROUP:", group_name)
    print("=" * 100)

    for target in features:

        predictors = [
            col for col in features
            if col != target
        ]

        y = Xz[target].to_numpy()

        candidate_results = []

        # Test subsets of 1, 2, and 3 predictors
        max_size = min(
            MAX_SUBSET_SIZE,
            len(predictors)
        )

        for subset_size in range(1, max_size + 1):

            for subset in combinations(
                predictors,
                subset_size
            ):

                A = Xz[list(subset)].to_numpy()

                result = evaluate_linear_identity(
                    y,
                    A,
                    list(subset)
                )

                row = {
                    "Group": group_name,
                    "Target": target,
                    "Model_Type": "Sparse",
                    "Predictor_Count": len(subset),
                    "Predictors": " + ".join(subset),
                    **{
                        key: value
                        for key, value in result.items()
                        if key != "Coefficients"
                    }
                }

                candidate_results.append({
                    "row": row,
                    "coefficients": result["Coefficients"]
                })

        # Rank sparse candidates
        candidate_results.sort(
            key=lambda item: item["row"]["Normalized_RMSE"]
        )

        # ====================================================
        # 5. FULL MULTIVARIABLE RECONSTRUCTION
        # ====================================================

        A_full = Xz[predictors].to_numpy()

        full_result = evaluate_linear_identity(
            y,
            A_full,
            predictors
        )

        full_row = {
            "Group": group_name,
            "Target": target,
            "Model_Type": "Full",
            "Predictor_Count": len(predictors),
            "Predictors": " + ".join(predictors),
            **{
                key: value
                for key, value in full_result.items()
                if key != "Coefficients"
            }
        }

        candidate_results.append({
            "row": full_row,
            "coefficients": full_result["Coefficients"]
        })

        # Save all results
        for item in candidate_results:
            all_results.append(item["row"])

        # ====================================================
        # 6. PRINT BEST MODELS
        # ====================================================

        print(f"\nTARGET: {target}")

        print("\nBest sparse models:")

        for item in candidate_results[:TOP_RESULTS_PER_TARGET]:

            row = item["row"]

            print(
                f"\nPredictors: {row['Predictors']}\n"
                f"R2: {row['R2']:.12f}\n"
                f"Normalized RMSE: "
                f"{row['Normalized_RMSE']:.12e}\n"
                f"MAE: {row['MAE']:.12e}\n"
                f"Maximum error: {row['Max_Abs_Error']:.12e}\n"
                f"Exact-row fraction: "
                f"{row['Exact_Row_Fraction']:.6%}\n"
                f"Numerical rank: {row['Rank']}"
            )

            print("Standardized coefficients:")

            for name, coefficient in (
                item["coefficients"].items()
            ):
                print(
                    f"  {name}: {coefficient:.12e}"
                )

        # Save the best sparse model and full model
        best_sparse = candidate_results[0]

        best_models.append({
            "Group": group_name,
            "Target": target,
            "Best_Sparse_Predictors":
                best_sparse["row"]["Predictors"],
            "Best_Sparse_R2":
                best_sparse["row"]["R2"],
            "Best_Sparse_Normalized_RMSE":
                best_sparse["row"]["Normalized_RMSE"],
            "Best_Sparse_Max_Error":
                best_sparse["row"]["Max_Abs_Error"],
            "Best_Sparse_Exact_Fraction":
                best_sparse["row"]["Exact_Row_Fraction"],
            "Full_Model_R2":
                full_result["R2"],
            "Full_Model_Normalized_RMSE":
                full_result["Normalized_RMSE"],
            "Full_Model_Max_Error":
                full_result["Max_Abs_Error"],
            "Full_Model_Exact_Fraction":
                full_result["Exact_Row_Fraction"]
        })

# ============================================================
# 7. CONSOLIDATED RESULTS
# ============================================================

identity_results = pd.DataFrame(all_results)

best_identity_results = pd.DataFrame(best_models)

print("\n" + "=" * 100)
print("BEST LINEAR IDENTITIES BY TARGET")
print("=" * 100)

print(
    best_identity_results.to_string(index=False)
)

# ============================================================
# 8. RANK THE STRONGEST IDENTITIES
# ============================================================

print("\n" + "=" * 100)
print("STRONGEST SPARSE LINEAR IDENTITIES")
print("=" * 100)

sparse_results = identity_results[
    identity_results["Model_Type"] == "Sparse"
].copy()

sparse_results = sparse_results.sort_values(
    "Normalized_RMSE",
    ascending=True
)

print(
    sparse_results.head(30).to_string(index=False)
)

# ============================================================
# 9. FLAG POTENTIAL NUMERICAL IDENTITIES
# ============================================================

print("\n" + "=" * 100)
print("POTENTIAL NEAR-EXACT IDENTITIES")
print("=" * 100)

potential_identities = identity_results[
    (identity_results["Normalized_RMSE"] < 1e-7) &
    (identity_results["Max_Abs_Error"] < 1e-6)
].copy()

if potential_identities.empty:

    print("No identities passed the selected numerical thresholds.")

else:

    print(
        potential_identities.sort_values(
            "Normalized_RMSE"
        ).to_string(index=False)
    )
```

```python
feature_groups = {
    "Voice_Call": [
        "plcd_vce_Mean",
        "attempt_Mean",
        "comp_vce_Mean",
        "complete_Mean",
        "peak_vce_Mean",
        "opk_vce_Mean"
    ],

    "Data_Call": [
        "plcd_dat_Mean",
        "comp_dat_Mean",
        "blck_dat_Mean",
        "drop_dat_Mean",
        "drop_blk_Mean",
        "unan_dat_Mean",
        "opk_dat_Mean",
        "peak_dat_Mean"
    ],

    "Dropped_Blocked_Calls": [
        "drop_blk_Mean",
        "blck_vce_Mean",
        "drop_vce_Mean",
        "blck_dat_Mean",
        "drop_dat_Mean"
    ]
}


# ============================================================
# 2. SPARSE RELATIONSHIPS FROM PREVIOUS ANALYSIS
# ============================================================

sparse_models = {
    "Voice_Call": {
        "plcd_vce_Mean": [
            "attempt_Mean", "comp_vce_Mean", "complete_Mean"
        ],
        "attempt_Mean": [
            "plcd_vce_Mean", "comp_vce_Mean", "complete_Mean"
        ],
        "comp_vce_Mean": [
            "plcd_vce_Mean", "attempt_Mean", "complete_Mean"
        ],
        "complete_Mean": [
            "plcd_vce_Mean", "attempt_Mean", "comp_vce_Mean"
        ],
    },

    "Data_Call": {
        "plcd_dat_Mean": [
            "comp_dat_Mean", "blck_dat_Mean", "drop_dat_Mean"
        ],
        "comp_dat_Mean": [
            "opk_dat_Mean", "peak_dat_Mean"
        ],
        "blck_dat_Mean": [
            "plcd_dat_Mean", "comp_dat_Mean", "drop_dat_Mean"
        ],
        "drop_dat_Mean": [
            "plcd_dat_Mean", "comp_dat_Mean", "blck_dat_Mean"
        ],
        "opk_dat_Mean": [
            "comp_dat_Mean", "peak_dat_Mean"
        ],
        "peak_dat_Mean": [
            "comp_dat_Mean", "opk_dat_Mean"
        ]
    },

    "Dropped_Blocked_Calls": {
        "drop_blk_Mean": [
            "blck_vce_Mean", "drop_vce_Mean", "blck_dat_Mean"
        ],
        "blck_vce_Mean": [
            "drop_blk_Mean", "drop_vce_Mean", "blck_dat_Mean"
        ],
        "drop_vce_Mean": [
            "drop_blk_Mean", "blck_vce_Mean", "blck_dat_Mean"
        ],
        "blck_dat_Mean": [
            "drop_blk_Mean", "blck_vce_Mean", "drop_vce_Mean"
        ],
        "drop_dat_Mean": [
            "drop_blk_Mean", "blck_vce_Mean", "drop_vce_Mean"
        ]
    }
}


# ============================================================
# 3. PREPARE DATA
# ============================================================

all_features = sorted(set(
    feature
    for group in feature_groups.values()
    for feature in group
))

missing_columns = [
    col for col in all_features
    if col not in telecom.columns
]

if missing_columns:
    raise ValueError(f"Missing columns: {missing_columns}")

# Keep original row indices for tracing unusual observations.
X_raw = telecom[all_features].apply(
    pd.to_numeric, errors="coerce"
).replace([np.inf, -np.inf], np.nan)

# Median imputation for diagnostic purposes only.
# This does not modify telecom.
X_clean = X_raw.copy()

X_clean = X_clean.fillna(X_clean.median())

# Remove any columns that remain entirely missing.
if X_clean.isna().any().any():
    raise ValueError("Some columns remain missing after median imputation.")


# ============================================================
# 4. RAW-SCALE RECONSTRUCTION FUNCTION
# ============================================================

def raw_reconstruction(
    data,
    target,
    predictors,
    group_name,
    model_type
):

    y = data[target].to_numpy(dtype=np.float64)
    X = data[predictors].to_numpy(dtype=np.float64)

    # Include intercept.
    X_design = np.column_stack([
        np.ones(len(X)),
        X
    ])

    # Least-squares solution.
    # In collinear systems, this returns a minimum-norm solution.
    coefficients, residuals, rank, singular_values = np.linalg.lstsq(
        X_design,
        y,
        rcond=None
    )

    y_pred = X_design @ coefficients
    errors = y - y_pred
    abs_errors = np.abs(errors)

    sse = np.sum(errors ** 2)
    sst = np.sum((y - np.mean(y)) ** 2)

    r2 = 1 - (sse / sst) if sst > 0 else np.nan
    rmse = np.sqrt(np.mean(errors ** 2))
    mae = np.mean(abs_errors)

    target_std = np.std(y)
    normalized_rmse = (
        rmse / target_std if target_std > 0 else np.nan
    )

    # Numerical tolerance: scale-aware, not a business-rule tolerance.
    tolerance = 1e-8 * np.maximum(1.0, np.abs(y))
    near_exact_fraction = np.mean(abs_errors <= tolerance)

    # Residual distribution.
    residual_quantiles = np.quantile(
        abs_errors,
        [0.50, 0.90, 0.95, 0.99, 0.999, 1.00]
    )

    # Coefficient output.
    coefficient_dict = {
        "Intercept": coefficients[0]
    }

    for predictor, coefficient in zip(
        predictors, coefficients[1:]
    ):
        coefficient_dict[predictor] = coefficient

    # Equation text.
    equation = f"{target} = {coefficients[0]:.12g}"

    for predictor, coefficient in zip(
        predictors, coefficients[1:]
    ):
        sign = "+" if coefficient >= 0 else "-"
        equation += (
            f" {sign} {abs(coefficient):.12g}*{predictor}"
        )

    # Worst reconstruction errors.
    worst_positions = np.argsort(abs_errors)[-10:][::-1]

    worst_rows = pd.DataFrame({
        "Original_Index": data.index[worst_positions],
        "Actual": y[worst_positions],
        "Predicted": y_pred[worst_positions],
        "Residual": errors[worst_positions],
        "Absolute_Error": abs_errors[worst_positions]
    })

    metrics = {
        "Group": group_name,
        "Model_Type": model_type,
        "Target": target,
        "Predictor_Count": len(predictors),
        "N": len(y),
        "R2": r2,
        "RMSE": rmse,
        "Normalized_RMSE": normalized_rmse,
        "MAE": mae,
        "Median_Abs_Error": residual_quantiles[0],
        "P90_Abs_Error": residual_quantiles[1],
        "P95_Abs_Error": residual_quantiles[2],
        "P99_Abs_Error": residual_quantiles[3],
        "P99.9_Abs_Error": residual_quantiles[4],
        "Maximum_Abs_Error": residual_quantiles[5],
        "Near_Exact_Fraction": near_exact_fraction,
        "Numerical_Rank": rank,
        "Design_Matrix_Columns": X_design.shape[1]
    }

    return {
        "metrics": metrics,
        "coefficients": coefficient_dict,
        "equation": equation,
        "worst_rows": worst_rows,
        "predictions": y_pred,
        "residuals": errors,
        "singular_values": singular_values
    }


# ============================================================
# 5. RUN FULL AND SPARSE RECONSTRUCTIONS
# ============================================================

results = {}
metrics_list = []

for group_name, features in feature_groups.items():

    print("\n" + "=" * 100)
    print(f"GROUP: {group_name}")
    print("=" * 100)

    for target in features:

        predictors_full = [
            col for col in features
            if col != target
        ]

        # --------------------------------------------
        # Full within-group model
        # --------------------------------------------

        full_result = raw_reconstruction(
            data=X_clean,
            target=target,
            predictors=predictors_full,
            group_name=group_name,
            model_type="Full"
        )

        results[(group_name, target, "Full")] = full_result
        metrics_list.append(full_result["metrics"])

        # --------------------------------------------
        # Sparse model
        # --------------------------------------------

        predictors_sparse = sparse_models.get(
            group_name, {}
        ).get(target)

        if predictors_sparse:

            sparse_result = raw_reconstruction(
                data=X_clean,
                target=target,
                predictors=predictors_sparse,
                group_name=group_name,
                model_type="Sparse"
            )

            results[(group_name, target, "Sparse")] = sparse_result
            metrics_list.append(sparse_result["metrics"])


# ============================================================
# 6. DISPLAY RAW-SCALE EQUATIONS
# ============================================================

print("\n" + "=" * 100)
print("RAW-SCALE RECONSTRUCTION EQUATIONS")
print("=" * 100)

for key, result in results.items():

    print(f"\n{key[0]} | {key[1]} | {key[2]}")
    print("-" * 100)

    print(result["equation"])

    print("\nCoefficients:")
    for name, value in result["coefficients"].items():
        print(f"{name:25s}: {value: .12g}")


# ============================================================
# 7. DISPLAY RESIDUAL DIAGNOSTICS
# ============================================================

print("\n" + "=" * 100)
print("RESIDUAL DIAGNOSTICS")
print("=" * 100)

for key, result in results.items():

    print(f"\n{key[0]} | {key[1]} | {key[2]}")
    print("-" * 100)

    m = result["metrics"]

    print(f"R2:                  {m['R2']:.12f}")
    print(f"RMSE:                {m['RMSE']:.12g}")
    print(f"Normalized RMSE:     {m['Normalized_RMSE']:.12g}")
    print(f"MAE:                 {m['MAE']:.12g}")
    print(f"Median |Residual|:   {m['Median_Abs_Error']:.12g}")
    print(f"P90 |Residual|:      {m['P90_Abs_Error']:.12g}")
    print(f"P95 |Residual|:      {m['P95_Abs_Error']:.12g}")
    print(f"P99 |Residual|:      {m['P99_Abs_Error']:.12g}")
    print(f"P99.9 |Residual|:    {m['P99.9_Abs_Error']:.12g}")
    print(f"Maximum error:       {m['Maximum_Abs_Error']:.12g}")
    print(f"Near-exact fraction: {m['Near_Exact_Fraction']:.6%}")
    print(
        f"Numerical rank:      "
        f"{m['Numerical_Rank']}/{m['Design_Matrix_Columns']}"
    )

    print("\nWorst 10 observations:")
    print(result["worst_rows"].to_string(index=False))


# ============================================================
# 8. SUMMARY TABLE
# ============================================================

reconstruction_summary = pd.DataFrame(metrics_list)

reconstruction_summary = reconstruction_summary.sort_values(
    ["Group", "Target", "Model_Type"]
).reset_index(drop=True)

print("\n" + "=" * 100)
print("RECONSTRUCTION SUMMARY")
print("=" * 100)

print(
    reconstruction_summary[
        [
            "Group",
            "Target",
            "Model_Type",
            "Predictor_Count",
            "R2",
            "Normalized_RMSE",
            "MAE",
            "P99_Abs_Error",
            "Maximum_Abs_Error",
            "Near_Exact_Fraction",
            "Numerical_Rank"
        ]
    ].to_string(index=False)
)
```

Feature elimination and Modelling

```python
lr = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler()),
    ("model", LogisticRegression(
        penalty="l2",
        C=1.0,
        solver="lbfgs",
        max_iter=3000,
        tol=1e-4,
        random_state=42
    ))
])


extra_trees = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("transformer", QuantileTransformer(
        n_quantiles=1000,
        output_distribution="normal",
        random_state=42
    )),
    ("model", ExtraTreesClassifier(
        n_estimators=500,
        criterion="gini",
        max_depth=None,
        min_samples_split=5,
        min_samples_leaf=3,
        max_features="sqrt",
        bootstrap=False,
        class_weight=None,
        n_jobs=1,
        random_state=42
    ))
])



rfr = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("transformer", QuantileTransformer(
        n_quantiles=1000,
        output_distribution="normal",
        random_state=42
    )),
    ("model", RandomForestClassifier(
        n_estimators=500,
        criterion="gini",
        max_depth=None,
        min_samples_split=5,
        min_samples_leaf=3,
        max_features="sqrt",
        bootstrap=True,
        max_samples=0.8,
        class_weight=None,
        n_jobs=1,
        random_state=42
    ))
])




xgb = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("transformer", QuantileTransformer(
        n_quantiles=1000,
        output_distribution="normal",
        random_state=42
    )),
    ("model", XGBClassifier(
        objective="binary:logistic",
        eval_metric="logloss",
        n_estimators=700,
        learning_rate=0.03,
        max_depth=4,
        min_child_weight=5,
        subsample=0.8,
        colsample_bytree=0.8,
        gamma=0.1,
        reg_alpha=0.1,
        reg_lambda=5.0,
        max_bin=256,
        tree_method="hist",
        n_jobs=1,
        random_state=42
    ))
])




lgbm = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("transformer", QuantileTransformer(
        n_quantiles=1000,
        output_distribution="normal",
        random_state=42
    )),
    ("model", LGBMClassifier(
        objective="binary",
        metric="binary_logloss",
        n_estimators=700,
        learning_rate=0.03,
        num_leaves=15,
        max_depth=6,
        min_child_samples=100,
        min_split_gain=0.0,
        subsample=0.8,
        subsample_freq=1,
        colsample_bytree=0.8,
        reg_alpha=0.1,
        reg_lambda=5.0,
        verbosity=-1,
        n_jobs=1,
        random_state=42
    ))
])




catboost = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("transformer", QuantileTransformer(
        n_quantiles=1000,
        output_distribution="normal",
        random_state=42
    )),
    ("model", CatBoostClassifier(
        loss_function="Logloss",
        eval_metric="AUC",
        iterations=800,
        learning_rate=0.03,
        depth=6,
        l2_leaf_reg=5.0,
        random_strength=1.0,
        border_count=128,
        bootstrap_type="Bayesian",
        bagging_temperature=1.0,
        allow_writing_files=False,
        thread_count=1,
        random_seed=42,
        verbose=0
    ))
])
```

```python
target = "churn"

X = (
    telecom
    .select_dtypes(include=np.number)
    .drop(columns=[target, "Customer_ID"], errors="ignore")
    .replace([np.inf, -np.inf], np.nan)
)

y = telecom[target].astype(int) 


features_A = X.columns.tolist()

# B: Aggregate-reduced feature set
features_B = [
    col for col in features_A
    if col not in [
        "plcd_dat_Mean",
        "drop_blk_Mean"
    ]
]

# C: Dependency-reduced feature set
# Retain plcd_vce, peak_vce, and opk_vce
# Remove highly redundant voice-call predictors.

features_C = [
    col for col in features_B
    if col not in [
        "attempt_Mean",
        "comp_vce_Mean",
        "complete_Mean"
    ]
]

feature_sets = {
    "A_Full": features_A,
    "B_Aggregate_Reduced": features_B,
    "C_Dependency_Reduced": features_C
}
```

```python
X_train_full, X_test_full, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

```python
models = {
    "LogisticRegression": lr,
    "ExtraTrees": extra_trees,
    "RandomForest": rfr,
    "XGBoost": xgb,
    "LightGBM": lgbm,
    "CatBoost": catboost
}

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

mlflow.set_experiment(
    "Telecom_Feature_Set_Robustness"
)

results = []

# ============================================================
# 6. CROSS-VALIDATION EXPERIMENT
# ============================================================

for feature_name, feature_columns in feature_sets.items():

    X_train = X_train_full[feature_columns]

    print("\n" + "=" * 85)
    print(f"FEATURE SET: {feature_name}")
    print(f"FEATURE COUNT: {len(feature_columns)}")
    print("=" * 85)

    for model_name, model in models.items():

        print(f"\nTraining {model_name}...")

        fold_metrics = []

        with mlflow.start_run(
            run_name=f"{feature_name}_{model_name}"
        ):

            mlflow.log_params({
                "feature_set": feature_name,
                "model": model_name,
                "n_features": len(feature_columns),
                "cv_folds": 5,
                "test_size": 0.20,
                "random_state": 42
            })

            for fold, (train_idx, val_idx) in enumerate(
                cv.split(X_train, y_train),
                start=1
            ):

                X_fold_train = X_train.iloc[train_idx]
                X_fold_val = X_train.iloc[val_idx]

                y_fold_train = y_train.iloc[train_idx]
                y_fold_val = y_train.iloc[val_idx]

                estimator = clone(model)

                estimator.fit(
                    X_fold_train,
                    y_fold_train
                )

                # Predicted probabilities
                y_prob = estimator.predict_proba(
                    X_fold_val
                )[:, 1]

                # Classification threshold
                y_pred = (y_prob >= 0.5).astype(int)

                # Confusion matrix
                tn, fp, fn, tp = confusion_matrix(
                    y_fold_val,
                    y_pred,
                    labels=[0, 1]
                ).ravel()

                # Evaluation metrics
                metrics = {
                    "ROC_AUC": roc_auc_score(
                        y_fold_val, y_prob
                    ),

                    "PR_AUC": average_precision_score(
                        y_fold_val, y_prob
                    ),

                    "MCC": matthews_corrcoef(
                        y_fold_val, y_pred
                    ),

                    "Sensitivity": recall_score(
                        y_fold_val,
                        y_pred,
                        zero_division=0
                    ),

                    "Specificity": (
                        tn / (tn + fp)
                        if (tn + fp) > 0 else 0
                    ),

                    "Brier_Score": brier_score_loss(
                        y_fold_val, y_prob
                    )
                }

                fold_metrics.append(metrics)

                # Log fold metrics
                mlflow.log_metrics({
                    f"{metric}_fold_{fold}": value
                    for metric, value in metrics.items()
                })

                print(
                    f"Fold {fold} | "
                    f"ROC-AUC: {metrics['ROC_AUC']:.4f} | "
                    f"PR-AUC: {metrics['PR_AUC']:.4f} | "
                    f"MCC: {metrics['MCC']:.4f} | "
                    f"Sensitivity: {metrics['Sensitivity']:.4f} | "
                    f"Specificity: {metrics['Specificity']:.4f}"
                )

                # Memory cleanup
                del estimator
                del X_fold_train, X_fold_val
                del y_fold_train, y_fold_val
                del y_prob, y_pred

                gc.collect()

            # ====================================================
            # 7. AGGREGATE CV METRICS
            # ====================================================

            fold_df = pd.DataFrame(fold_metrics)

            summary = {}

            for metric in fold_df.columns:

                mean_score = fold_df[metric].mean()
                std_score = fold_df[metric].std()

                summary[f"{metric}_mean"] = mean_score
                summary[f"{metric}_std"] = std_score

                mlflow.log_metric(
                    f"{metric}_mean",
                    mean_score
                )

                mlflow.log_metric(
                    f"{metric}_std",
                    std_score
                )

            results.append({
                "Feature_Set": feature_name,
                "Model": model_name,
                "N_Features": len(feature_columns),
                **summary
            })

            print("\nCV SUMMARY:")
            print(fold_df.mean().round(4))

            del fold_df
            gc.collect()

    del X_train
    gc.collect()

# ============================================================
# 8. CROSS-VALIDATION RESULTS
# ============================================================

results_df = pd.DataFrame(results)

results_df = results_df.sort_values(
    by=["MCC_mean", "ROC_AUC_mean"],
    ascending=False
).reset_index(drop=True)

print("\n" + "=" * 100)
print("CROSS-VALIDATION RESULTS")
print("=" * 100)

print(results_df.round(4).to_string(index=False))

# ============================================================
# 9. FEATURE SET COMPARISON
# ============================================================

feature_set_comparison = (
    results_df
    .groupby("Feature_Set")
    .agg(
        Mean_ROC_AUC=("ROC_AUC_mean", "mean"),
        Mean_PR_AUC=("PR_AUC_mean", "mean"),
        Mean_MCC=("MCC_mean", "mean"),
        Mean_Sensitivity=("Sensitivity_mean", "mean"),
        Mean_Specificity=("Specificity_mean", "mean"),
        Mean_Brier=("Brier_Score_mean", "mean")
    )
    .sort_values("Mean_MCC", ascending=False)
)

print("\nFEATURE SET COMPARISON")
print(feature_set_comparison.round(4))

# ============================================================
# 10. SELECT BEST MODEL PER FEATURE SET
# ============================================================

best_per_feature_set = (
    results_df
    .sort_values(
        ["MCC_mean", "ROC_AUC_mean"],
        ascending=False
    )
    .groupby("Feature_Set")
    .head(1)
    .reset_index(drop=True)
)

print("\nBEST MODEL PER FEATURE SET")
print(
    best_per_feature_set[
        [
            "Feature_Set",
            "Model",
            "N_Features",
            "ROC_AUC_mean",
            "PR_AUC_mean",
            "MCC_mean",
            "Sensitivity_mean",
            "Specificity_mean",
            "Brier_Score_mean"
        ]
    ].round(4).to_string(index=False)
)

# ============================================================
# 11. FINAL HELD-OUT TEST EVALUATION
# ============================================================

test_results = []

for _, best_row in best_per_feature_set.iterrows():

    feature_name = best_row["Feature_Set"]
    model_name = best_row["Model"]

    feature_columns = feature_sets[feature_name]

    print("\n" + "=" * 85)
    print(f"FINAL TEST EVALUATION: {feature_name}")
    print(f"SELECTED MODEL: {model_name}")
    print("=" * 85)

    X_train = X_train_full[feature_columns]
    X_test = X_test_full[feature_columns]

    final_model = clone(models[model_name])

    # Fit on the complete training partition
    final_model.fit(
        X_train,
        y_train
    )

    # Evaluate on untouched test partition
    y_prob = final_model.predict_proba(
        X_test
    )[:, 1]

    y_pred = (y_prob >= 0.5).astype(int)

    tn, fp, fn, tp = confusion_matrix(
        y_test,
        y_pred,
        labels=[0, 1]
    ).ravel()

    test_metrics = {
        "Feature_Set": feature_name,
        "Model": model_name,
        "N_Features": len(feature_columns),

        "ROC_AUC": roc_auc_score(
            y_test, y_prob
        ),

        "PR_AUC": average_precision_score(
            y_test, y_prob
        ),

        "MCC": matthews_corrcoef(
            y_test, y_pred
        ),

        "Sensitivity": recall_score(
            y_test,
            y_pred,
            zero_division=0
        ),

        "Specificity": (
            tn / (tn + fp)
            if (tn + fp) > 0 else 0
        ),

        "Brier_Score": brier_score_loss(
            y_test, y_prob
        ),

        "TP": tp,
        "TN": tn,
        "FP": fp,
        "FN": fn
    }

    test_results.append(test_metrics)

    print(pd.Series(test_metrics).round(4))

    # Log final test evaluation
    with mlflow.start_run(
        run_name=f"TEST_{feature_name}_{model_name}"
    ):

        mlflow.log_params({
            "feature_set": feature_name,
            "model": model_name,
            "evaluation": "held_out_test",
            "n_features": len(feature_columns)
        })

        mlflow.log_metrics({
            key: value
            for key, value in test_metrics.items()
            if isinstance(value, (int, float, np.integer, np.floating))
            and key != "N_Features"
        })

    del final_model, X_train, X_test
    del y_prob, y_pred
    gc.collect()

# ============================================================
# 12. FINAL TEST RESULTS
# ============================================================

test_results_df = pd.DataFrame(test_results)

print("\n" + "=" * 100)
print("FINAL HELD-OUT TEST RESULTS")
print("=" * 100)

print(test_results_df.round(4).to_string(index=False))

# ============================================================
# 13. SAVE RESULTS
# ============================================================

results_df.to_csv(
    "telecom_cv_results.csv",
    index=False
)

test_results_df.to_csv(
    "telecom_test_results.csv",
    index=False
)

feature_set_comparison.to_csv(
    "telecom_feature_set_comparison.csv"
)

print("\nAll experiments completed.")
print("Results saved successfully.")
```
